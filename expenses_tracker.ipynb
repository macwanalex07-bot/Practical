```python
import pandas as pd
import mathplotlib as plt
import seaborn as sns
import os
```

```python
def generate_charts(expenses_df):
        spending_by_category = expenses_df.groupby('Category')['Amount'].sum().sort_values(ascending=False)
        expenses_df['Date']=pd.to_datetime(expenses_df['Date'])
        spending_over_time=expenses_df.set_index('Date').resample('M')['Amount'].sum()
        
        plt.style.use('seaborn-v0_8-whitegrid')

        plt.figure(figsize=12,7)
        sns.barplot(x=spending_by_category.index, y=spending_by_category.values,palette='viridis')
        plt.title('Total Expenses by Category',fontsize=18,fontweight='bold')
        plt.xlabel('Category',fontsize=14)
        plt.ylabel('Total Amount ($)',fontsize=14)
        plt.xticks(rotation=45,ha='right')
        plt.tight_layout()
        plt.show()


        plt.figure(figsize=(10,10))
        plt.pie(spending_by_category, labels=spending_by_category.index, autopct='%1.1f%%',startangle=140,colors=sns.color_palette('pastel'))
        plt.title('Proportional spending Distribution by Category',fontsize=18,fontweight='bold')
        plt.axis('equal')
        plt.tight_layout()
        plt.show()


        plt.figure(figsize=12,7))
        sns.lineplot(x=spending_over_time.index, y=spending_over_time.values,marker='o',color='royalblue')
        plt.title('Monthly Spending Trends Over Time',fontsize=18,fontweight='bold')
        plt.xlabel('Date',fontsize=14)
        plt.ylabel('Total Monthly Spending($)',fontsize=14)
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()




        plt.figure(figsize=(12,7)
        sns.hisplot(expenses_df['Amount'], bins=20, kde=True, color='darkcyan')
        plt.title('Frequency of Expense Amounts', fontsize=18, fontweight='bold')
        plt.xlabel('Frequency of Expense Amount($)',fontsize=14)
        plt.ylabel('Frequency', fontsize=14)
        plt.tight_layout()
        plt.show()
        
```
